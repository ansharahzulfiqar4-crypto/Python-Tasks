```python
# Task 1: Cinema Ticket Price


name = input("Enter your name")
age = int(input("Enter your age:"))

if age < 5 or age > 100:
  ticket_price = "Free"
elif age <= 12:
  ticket_price = "Rs. 300"
elif age <= 59:
  ticket_price = "Rs. 700"
else:
  ticket_price = "Rs. 400"

print("Customer:", name)
print("Ticket price",ticket_price)
```

```python
## Task 2: Temperature Converter

celsius = float(input("Enter temperature in Celsius:"))

fahrenheit = (celsius * 9/5) + 32

if celsius < 10:
  weather = "Very cold"
elif celsius <= 19:
  weather = "Cold"
elif celsius <= 29:
  weather = "Normal"
else:
  weather = "hot"

print("Temperature in Fahrenheit:", fahrenheit)
print("Weather Condition:", weather)
```

```python
## Task 3: Mobile Recharge Calculator

mobile_number = input("Enter mobile number")
recharge_amount = float(input("Enter recharge amount:"))

# calculate 3% tax
tax = recharge_amount * 3 / 100

#Calculate final balance
final_balance = recharge_amount -  tax

#Display result
print("Recharge Details")
print("Mobile Number:",mobile_number)
print("Recharge  Aamount:",recharge_amount)
print("Tax (3%):", tax)
print("Final Received Balance:",final_balance)
```

```python
# Intermediate - Level Task

#Task 6: Grocery Shopping Bill
total_products = int(input("Enter total products: "))

total_bill = 0

for i in range(total_products):
    print(f"\nProduct {i + 1}")

    name = input("Enter product name: ")
    price = float(input("Enter price: Rs. "))
    quantity = int(input("Enter quantity: "))

    subtotal = price * quantity
    total_bill += subtotal

    print("Subtotal:", subtotal)

if total_bill >= 5000:
    discount = total_bill * 0.10
else:
    discount = 0

final_bill = total_bill - discount

print("\n===== GROCERY BILL =====")
print("Total Bill: Rs.", total_bill)
print("Discount: Rs.", discount)
print("Final Bill: Rs.", final_bill)
```

```python
#Task 7: Employee Weekely Salary
name = input("Enter employee name: ")
hours = float(input("Enter total working hours: "))
rate = float(input("Enter hourly rate: Rs. "))

if hours <= 40:
    normal_salary = hours * rate
    overtime_salary = 0
else:
    normal_salary = 40 * rate
    overtime_hours = hours - 40
    overtime_salary = overtime_hours * rate * 1.5

total_salary = normal_salary + overtime_salary

print("\n===== WEEKLY SALARY =====")
print("Employee Name:", name)
print("Normal Salary: Rs.", normal_salary)
print("Overtime Salary: Rs.", overtime_salary)
print("Total Weekly Salary: Rs.", total_salary)
```

```python
#Advanced Level Tasks
#Task 11: Personal Expense Tracker
 def final_report(expenses, budget):
    total = 0

    category_total = {
        "Food": 0,
        "Transport": 0,
        "Shopping": 0,
        "Bills": 0,
        "Other": 0
    }


    for expense in expenses:
        total = total + expense["amount"]
        category_total[expense["category"]] += expense["amount"]


    highest = expenses[0]

    for expense in expenses:
        if expense["amount"] > highest["amount"]:
            highest = expense

    print("\n========== EXPENSE REPORT ==========")
    print("Total Spending: Rs.", total)

    print("\nCategory Spending:")
    for category in category_total:
        print(category, ": Rs.", category_total[category])

    print("\nHighest Expense:")
    print("Name:", highest["name"])
    print("Category:", highest["category"])
    print("Amount: Rs.", highest["amount"])

    print("\nBudget: Rs.", budget)

    if total > budget:
        print("WARNING: Budget Exceeded!")
        print("Exceeded Amount: Rs.", total - budget)
    else:
        print("Budget is within limit.")
        print("Remaining Budget: Rs.", budget - total)




expenses = []

categories = ["Food", "Transport", "Shopping", "Bills", "Other"]

total_expenses = int(input("Enter total expenses: "))

for i in range(total_expenses):
    print("\nExpense", i + 1)

    name = input("Enter expense name: ")

    print("Categories:", categories)
    category = input("Enter category: ")

    amount = float(input("Enter amount: Rs. "))

    expense = {
        "name": name,
        "category": category,
        "amount": amount
    }

    expenses.append(expense)

budget = float(input("\nEnter your budget: Rs. "))

final_report(expenses, budget)
```

```python
#Task 12: Store Inventory Manager


inventory = {}

def add_product():
    name = input("Enter product name: ")

    if name in inventory:
        print("Product already exists!")
        return

    price = float(input("Enter product price: Rs. "))
    quantity = int(input("Enter available quantity: "))

    inventory[name] = {
        "price": price,
        "quantity": quantity
    }

    print("Product added successfully!")


def sell_product():
    name = input("Enter product name: ")

    if name not in inventory:
        print("Product not found!")
        return

    quantity = int(input("Enter quantity to sell: "))

    if quantity > inventory[name]["quantity"]:
        print("WARNING: Insufficient stock!")
    else:
        inventory[name]["quantity"] -= quantity
        print("Product sold successfully!")


def restock_product():
    name = input("Enter product name: ")

    if name not in inventory:
        print("Product not found!")
        return

    quantity = int(input("Enter quantity to add: "))

    inventory[name]["quantity"] += quantity

    print("Product restocked successfully!")


def search_product():
    name = input("Enter product name: ")

    if name in inventory:
        print("\nProduct Found!")
        print("Name:", name)
        print("Price: Rs.", inventory[name]["price"])
        print("Quantity:", inventory[name]["quantity"])
    else:
        print("Product not found!")


def display_inventory():
    if not inventory:
        print("Inventory is empty!")
        return

    print("\n========== INVENTORY ==========")

    total_value = 0

    for name, product in inventory.items():
        price = product["price"]
        quantity = product["quantity"]
        value = price * quantity

        print("\nProduct:", name)
        print("Price: Rs.", price)
        print("Quantity:", quantity)
        print("Product Value: Rs.", value)

        total_value += value

    print("\nTotal Inventory Value: Rs.", total_value)


def low_stock():
    print("\n========== LOW STOCK ==========")

    found = False

    for name, product in inventory.items():
        if product["quantity"] < 5:
            print(name, "- Quantity:", product["quantity"])
            found = True

    if not found:
        print("No low-stock products.")


while True:
    print("\n========== STORE INVENTORY ==========")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Display Low Stock")
    print("7. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()

    elif choice == "2":
        sell_product()

    elif choice == "3":
        restock_product()

    elif choice == "4":
        search_product()

    elif choice == "5":
        display_inventory()

    elif choice == "6":
        low_stock()

    elif choice == "7":
        print("Program ended. Thank you!")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
#Task 13: Library Book Management System
books = []


def add_book():
    book_id = input("Enter Book ID: ")


    for book in books:
        if book["id"] == book_id:
            print("Book ID already exists!")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")

    book = {
        "id": book_id,
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)

    print("Book added successfully!")


def display_books():
    if len(books) == 0:
        print("No books available!")
        return

    print("\n===== LIBRARY BOOKS =====")

    for book in books:
        if book["available"]:
            status = "Available"
        else:
            status = "Borrowed"

        print("\nBook ID:", book["id"])
        print("Title:", book["title"])
        print("Author:", book["author"])
        print("Status:", status)



def search_book():
    title = input("Enter book title to search: ")

    found = False

    for book in books:
        if book["title"].lower() == title.lower():
            print("\nBook Found!")
            print("Book ID:", book["id"])
            print("Title:", book["title"])
            print("Author:", book["author"])

            if book["available"]:
                print("Status: Available")
            else:
                print("Status: Borrowed")

            found = True

    if not found:
        print("Book not found!")


def borrow_book():
    book_id = input("Enter Book ID to borrow: ")

    for book in books:
        if book["id"] == book_id:

            if book["available"]:
                book["available"] = False
                print("Book borrowed successfully!")
            else:
                print("Book is already borrowed!")

            return

    print("Book ID not found!")


def return_book():
    book_id = input("Enter Book ID to return: ")

    for book in books:
        if book["id"] == book_id:

            if not book["available"]:
                book["available"] = True
                print("Book returned successfully!")
            else:
                print("This book is already available!")

            return

    print("Book ID not found!")


while True:

    print("\n===== LIBRARY MENU =====")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Borrow Book")
    print("5. Return Book")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_book()

    elif choice == "2":
        display_books()

    elif choice == "3":
        search_book()

    elif choice == "4":
        borrow_book()

    elif choice == "5":
        return_book()

    elif choice == "6":
        print("Library system closed.")
        break

    else:
        print("Invalid choice! Please try again.")
```

```python
# Task 14: Restaurant Ordering System
menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

orders = []


def display_menu():
    print("\n===== RESTAURANT MENU =====")

    for item, price in menu.items():
        print(item, "- Rs.", price)


def take_order():
    while True:
        display_menu()

        item = input("\nEnter item name (or 'done' to finish): ")

        if item.lower() == "done":
            break

        if item not in menu:
            print("Item is not available!")
            continue

        quantity = int(input("Enter quantity: "))

        subtotal = menu[item] * quantity

        order = {
            "item": item,
            "price": menu[item],
            "quantity": quantity,
            "subtotal": subtotal
        }

        orders.append(order)

        print("Item added to order!")


def display_receipt():
    print("\n========== RECEIPT ==========")

    total = 0

    for order in orders:
        print(
            order["item"],
            "x",
            order["quantity"],
            "= Rs.",
            order["subtotal"]
        )

        total += order["subtotal"]


    if total >= 3000:
        discount = total * 0.10
    else:
        discount = 0


    tax = total * 0.13

    final_bill = total + tax - discount

    print("-----------------------------")
    print("Total: Rs.", total)
    print("Discount: Rs.", discount)
    print("Tax (13%): Rs.", tax)
    print("Final Bill: Rs.", final_bill)
    print("=============================")



print("===== WELCOME TO RESTAURANT =====")

take_order()

if len(orders) > 0:
    display_receipt()
else:
    print("No items ordered.")

print("Thank you for visiting!")
```

```python
#Task 15:Clinic Appointment Manager
patients = []


def add_appointment():
    patient_id = input("Enter Patient ID: ")


    for patient in patients:
        if patient["id"] == patient_id:
            print("Patient ID already exists!")
            return

    name = input("Enter Patient Name: ")
    age = int(input("Enter Age: "))
    problem = input("Enter Disease/Problem: ")
    time = input("Enter Appointment Time: ")

    emergency = input("Is this an emergency? (yes/no): ")

    if emergency.lower() == "yes":
        emergency_status = True
    else:
        emergency_status = False

    patient = {
        "id": patient_id,
        "name": name,
        "age": age,
        "problem": problem,
        "time": time,
        "emergency": emergency_status,
        "cancelled": False
    }

    patients.append(patient)

    print("Appointment added successfully!")



def search_patient():
    patient_id = input("Enter Patient ID: ")

    for patient in patients:
        if patient["id"] == patient_id:
            print("\n===== PATIENT FOUND =====")
            print("Patient ID:", patient["id"])
            print("Name:", patient["name"])
            print("Age:", patient["age"])
            print("Problem:", patient["problem"])
            print("Appointment Time:", patient["time"])

            if patient["emergency"]:
                print("Emergency: Yes")
            else:
                print("Emergency: No")

            if patient["cancelled"]:
                print("Status: Cancelled")
            else:
                print("Status: Active")

            return

    print("Patient not found!")



def display_appointments():
    if len(patients) == 0:
        print("No appointments available!")
        return

    print("\n===== EMERGENCY PATIENTS =====")


    for patient in patients:
        if patient["emergency"] and not patient["cancelled"]:
            print(
                patient["id"],
                "-",
                patient["name"],
                "-",
                patient["time"]
            )

    print("\n===== NORMAL PATIENTS =====")


    for patient in patients:
        if not patient["emergency"] and not patient["cancelled"]:
            print(
                patient["id"],
                "-",
                patient["name"],
                "-",
                patient["time"]
            )


def cancel_appointment():
    patient_id = input("Enter Patient ID to cancel: ")

    for patient in patients:
        if patient["id"] == patient_id:

            if patient["cancelled"]:
                print("Appointment is already cancelled!")
            else:
                patient["cancelled"] = True
                print("Appointment cancelled successfully!")

            return

    print("Patient not found!")


def clinic_summary():
    normal_count = 0
    emergency_count = 0

    for patient in patients:
        if not patient["cancelled"]:
            if patient["emergency"]:
                emergency_count += 1
            else:
                normal_count += 1

    print("\n===== CLINIC SUMMARY =====")
    print("Normal Patients:", normal_count)
    print("Emergency Patients:", emergency_count)
    print("Total Active Patients:", normal_count + emergency_count)


while True:

    print("\n===== CLINIC APPOINTMENT MANAGER =====")
    print("1. Add Appointment")
    print("2. Search Patient")
    print("3. Display Appointments")
    print("4. Cancel Appointment")
    print("5. Display Clinic Summary")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_appointment()

    elif choice == "2":
        search_patient()

    elif choice == "3":
        display_appointments()

    elif choice == "4":
        cancel_appointment()

    elif choice == "5":
        clinic_summary()

    elif choice == "6":
        print("Clinic system closed.")
        break

    else:
        print("Invalid choice! Please try again.")
```
